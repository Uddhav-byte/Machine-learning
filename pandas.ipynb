handling missing cases

```python
df = pd.DataFrame({
    "Name": ["Aman", "Rahul", "Priya", "Neha"],
    "Age": [20, None, 20, 22],
    "Marks": [85, 90, None, 92]
})

print(df)
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
#count missing value per column
```

```python
print(df.isnull().sum().sum())
```

```python
df_clean = df.dropna()
#This removes rows containing at least one missing value by default.
#It does not change the original df unless you assign the result back or use an appropriate in-place operation.
```

```python
print(df)
```

Fill the missing value either by mean ,mode,median

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].median())
```

```python
print(df)
```

```python
df = pd.DataFrame({
    "Name": ["Aman", "Rahul", "Priya"],
    "Marks": [85, 90, 78]
})

df["Passed"] = df["Marks"] >= 40

df["Bonus_Marks"] = df["Marks"] + 5

print(df)
```

```python
df = df.rename(columns={"Marks": "Score"})
```

```python
print(df)
```
