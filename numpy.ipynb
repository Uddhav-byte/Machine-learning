```python
#implementation
import numpy as np

X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([20, 40, 60, 80, 100], dtype=float)
#X contains study hours.
#y contains actual exam scores.
w = 0.0
b = 0.0

predictions = w * X + b

print(predictions)
# [0. 0. 0. 0. 0.]

errors = predictions - y

cost = np.mean(errors ** 2)

print(cost)
#Gradient descent adjusts the weight and bias to reduce the cost.
```

```python

import numpy as np

# Training data
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([20, 40, 60, 80, 100], dtype=float)

# Initialize parameters
w = 0.0
b = 0.0

learning_rate = 0.01
m = len(X)

for epoch in range(1000):
    # 1. Predict all examples
    predictions = w * X + b

    # 2. Calculate errors
    errors = predictions - y

    # 3. Calculate gradients
    dw = (2 / m) * np.sum(errors * X)
    db = (2 / m) * np.sum(errors)

    # 4. Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db

# Final results
print("Weight:", w)
print("Bias:", b)
print("Predictions:", w * X + b)
print("Cost:", np.mean((w * X + b - y) ** 2))
```
